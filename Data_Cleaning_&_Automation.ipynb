```python
import numpy as np
import pandas as pd

df = pd.read_csv("/content/my_file (1).csv")
df.head()
```

```python
# Clean Column names
df.columns
```

```python
cleaned_columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
cleaned_columns = cleaned_columns.str.replace('year(s)', 'year')
cleaned_columns
```

```python
df.info()
```

```python
# Remove spaces in columns
for i in df.select_dtypes(include='object').columns:
  df[i] = df[i].str.strip()

df.head(10)
```

```python
# Normalizing NA values
na_values = ['na','n/a','none',' ','null','nan']
for i in df.columns:
  if df[i].dtype == 'object':
    df[i] = df[i].replace(to_replace=na_values,value=np.nan)
df.head(10)
```

```python
# Handling Missing Values
import pandas.api.types as pd_types
for col in df.columns:
  if pd_types.is_numeric_dtype(df[col]):
    df[col] = df[col].fillna(df[col].median())

  elif pd_types.is_object_dtype(df[col]):
    df[col] = df[col].fillna(df[col].mode()[0])

  else:
    df[col] = df[col].fillna(df[col].mean())

df.head(5)
```

```python
# Clean currency and related symbols in columns
for col in df.columns:
  if df[col].dtype == 'object':
    df[col] = df[col].str.replace('£','')
    df[col] = df[col].str.replace('$','')
    df[col] = df[col].str.replace('€','')

df
```

```python
df.to_csv("Cleaned.csv")
```
